```python
import pandas as pd
```

```python
games = pd.read_csv('./games.csv')
games.head()
```

```python
games.info()
```

```python
games = games[['GAME_DATE_EST', 'GAME_ID', 'GAME_STATUS_TEXT', 'TEAM_ID_home', 'TEAM_ID_away', 'PTS_home', 
               'PTS_away', 'HOME_TEAM_WINS']]
games.head()
```

```python
games['GAME_DATE_EST'] = pd.to_datetime(games['GAME_DATE_EST'])
#games.loc[:, 'GAME_DATE_EST'] = pd.to_datetime(games['GAME_DATE_EST'])
games.info()
```

```python
games['year'] = games['GAME_DATE_EST'].dt.year
games.head()
```

```python
# Cargar el conjunto de teams
teams = pd.read_csv('./teams.csv')
teams.head()
```

```python
teams = teams[['TEAM_ID', 'CITY', 'NICKNAME']]
teams.head()
```

```python
home_games = pd.merge(games, teams, left_on='TEAM_ID_home', right_on='TEAM_ID', how='inner')
home_games.head()
```

```python
home_games = home_games.rename(columns={
    'CITY': 'city_home',
    'NICKNAME': 'nickname_home'
})

home_games.head()
```

```python
both_games = pd.merge(home_games, teams, left_on='TEAM_ID_away', right_on='TEAM_ID', how='inner')
both_games.head()
```

```python
both_games = both_games.rename(columns={
    'CITY': 'city_away',
    'NICKNAME': 'nickname_away'
})

both_games.head()
```

```python
both_games = both_games.drop(columns=[
    'TEAM_ID_home', 'TEAM_ID_away', 'TEAM_ID_x', 'TEAM_ID_y'])

both_games.head()
```

```python
games = both_games.copy()
games.columns = games.columns.str.lower()
games.head()
```

```python
games['total_points'] = games['pts_home'] + games['pts_away']
games.head()
```

```python
games.nlargest(n=5, columns='total_points')
```

```python
games.nsmallest(n=5, columns='total_points')
```

## Graficos 

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
total = games['home_team_wins'].value_counts()
total
```

```python
plt.pie(total.values, labels=['Casa', 'Visitante'], autopct='%.0f%%')
plt.show()
```

```python
games.head()
```

```python
# Cantidad de puntos por equipo cuando juegan de local o de visita
plt.figure(figsize=(20, 6))
home_wins = games[games['home_team_wins'] == 1]
resultado = home_wins.groupby('nickname_home')['pts_home'].aggregate(np.mean).reset_index().sort_values('pts_home')

ax = sns.barplot(data=home_wins, x='nickname_home', y='pts_home', order=resultado['nickname_home'], errorbar=None)
ax.bar_label(ax.containers[-1], fmt='%.0f', label_type='center', rotation=45)
plt.xticks(rotation=75)
plt.show()
```

```python
plt.figure(figsize=(20, 6))
away_wins = games[games['home_team_wins'] == 0]
resultado = away_wins.groupby('nickname_away')['pts_away'].aggregate(np.mean).reset_index().sort_values('pts_away')

ax = sns.barplot(data=home_wins, x='nickname_away', y='pts_away', order=resultado['nickname_away'], errorbar=None)
ax.bar_label(ax.containers[-1], fmt='%.0f', label_type='center', rotation=45)
plt.xticks(rotation=75)
plt.show()
```
